```python
import pandas as pd
import numpy as np
import os  # To save the CSV file

# --- Configuration ---
N_ROWS = 10000
FILE_NAME = 'realistic_migraine_data.csv'
np.random.seed(42)  # For reproducible results

# --- 1. Initialize Dataset & Generate Demographic Features ---
data = {}

# Gender (0=Female, 1=Male): Skewed toward Female (~75% F : 25% M)
data['Gender'] = np.random.choice([0, 1], N_ROWS, p=[0.75, 0.25])
is_female = data['Gender'] == 0

# Age: Peak incidence around 35. Use a distribution centered around 37 (18–65 range).
base_age = np.random.normal(loc=37, scale=10, size=N_ROWS)
data['Age'] = np.clip(base_age, 18, 65).astype(int)

# Menstruation (0/1): Only applicable to females (0) of reproductive age (18–50).
# Assume ~20% of relevant female entries are experiencing hormonal triggers.
is_reproductive_age_female = (is_female) & (data['Age'] >= 18) & (data['Age'] <= 50)
menstruation_potential = np.random.choice([0, 1], N_ROWS, p=[0.80, 0.20])
data['Menstruation'] = np.where(is_reproductive_age_female, menstruation_potential, 0)

# --- 2. Lifestyle / Trigger Features ---

# Stress Level (1–10)
data['Stress_Level'] = np.clip(
    np.round(np.random.normal(loc=6.0, scale=2.5, size=N_ROWS)),
    1, 10
).astype(int)

# Sleep Duration (hours)
data['Sleep_Duration'] = np.clip(
    np.random.normal(loc=6.8, scale=1.3, size=N_ROWS),
    4.0, 9.5
).round(1)

# Number of meals (1–5), Water intake (1–5)
data['No_of_Meals'] = np.random.randint(2, 6, N_ROWS)
data['Water_Intake'] = np.random.randint(1, 6, N_ROWS)

# --- 3. Pain Severity Calculation (clinically inspired) ---

# Trigger-based contributions
stress_weight = data['Stress_Level'] / 10 * 3.0
sleep_weight = (9.5 - data['Sleep_Duration']) / 5.5 * 3.0  # less sleep → higher risk
menstrual_weight = data['Menstruation'] * 2.5
random_noise = np.random.uniform(1.0, 3.0, N_ROWS)

# Raw pain score
base_pain_score = stress_weight + sleep_weight + menstrual_weight + random_noise

# Clamp to 1–10 scale
data['Pain_Severity'] = np.clip(np.round(base_pain_score), 1, 10).astype(int)

# --- 4. Sensitivity Features (correlated with pain) ---

data['Light_Sensitivity'] = (
    ((data['Pain_Severity'] >= 6) & (np.random.rand(N_ROWS) < 0.85)) |
    ((data['Pain_Severity'] < 6) & (np.random.rand(N_ROWS) < 0.20))
).astype(int)

data['Noise_Sensitivity'] = (
    ((data['Pain_Severity'] >= 5) & (np.random.rand(N_ROWS) < 0.75)) |
    ((data['Pain_Severity'] < 5) & (np.random.rand(N_ROWS) < 0.15))
).astype(int)

# --- 5. Weather Features (one-hot encoded) ---
weather_options = ['Sunny', 'Cloudy', 'Rainy', 'Snowy']
weather_probs = [0.4, 0.35, 0.2, 0.05]

weather_choices = np.random.choice(weather_options, N_ROWS, p=weather_probs)
weather_df = pd.DataFrame(0, index=range(N_ROWS), columns=[f'Weather_{w}' for w in weather_options])

for i, choice in enumerate(weather_choices):
    weather_df.loc[i, f'Weather_{choice}'] = 1

# --- 6. Combine all data ---
df = pd.DataFrame(data)
df = pd.concat([df, weather_df], axis=1)

# --- 7. Migraine Score (0–100) ---

# Normalize features
norm_stress = df['Stress_Level'] / 10
norm_sleep = (9.5 - df['Sleep_Duration']) / 5.5
norm_menstruation = df['Menstruation']
norm_pain = df['Pain_Severity'] / 10
norm_light = df['Light_Sensitivity']
norm_noise = df['Noise_Sensitivity']

# Weighted factors (sum to 1.00)
weights = {
    'pain': 0.35,
    'stress': 0.20,
    'sleep': 0.15,
    'light': 0.10,
    'noise': 0.10,
    'menstruation': 0.10
}

migraine_score = (
    weights['pain'] * norm_pain +
    weights['stress'] * norm_stress +
    weights['sleep'] * norm_sleep +
    weights['light'] * norm_light +
    weights['noise'] * norm_noise +
    weights['menstruation'] * norm_menstruation
)

df['Migraine_Score'] = (migraine_score * 100).round(2)

# --- 8. Add Noise & Bias for Better Generalization ---

# --- A. Add Continuous Noise ---
# Jitter sleep duration (+/- 0.3 hours)
df['Sleep_Duration'] = df['Sleep_Duration'] + np.random.normal(0, 0.3, N_ROWS)
df['Sleep_Duration'] = df['Sleep_Duration'].clip(4, 9.5).round(1)

# Stress noise (+/- 1 point)
df['Stress_Level'] = df['Stress_Level'] + np.random.randint(-1, 2, N_ROWS)
df['Stress_Level'] = df['Stress_Level'].clip(1, 10)

# Pain severity noise (+/- 1 point)
df['Pain_Severity'] = df['Pain_Severity'] + np.random.choice([-1, 0, 1], N_ROWS, p=[0.1, 0.8, 0.1])
df['Pain_Severity'] = df['Pain_Severity'].clip(1, 10)

# --- B. Bias: Age-based underreporting of stress ---
young_mask = df['Age'] < 25
df.loc[young_mask, 'Stress_Level'] = (df.loc[young_mask, 'Stress_Level'] - 1).clip(1, 10)

# --- C. Bias: Weather impact on pain ---
df.loc[df['Weather_Rainy'] == 1, 'Pain_Severity'] += np.random.choice([0, 1], size=(df['Weather_Rainy'] == 1).sum(), p=[0.7, 0.3])
df['Pain_Severity'] = df['Pain_Severity'].clip(1, 10)

# --- D. Gender-based slight increase in migraine risk (real clinical pattern) ---
df.loc[df['Gender'] == 0, 'Migraine_Score'] *= 1.05
df['Migraine_Score'] = df['Migraine_Score'].clip(0, 100).round(2)

# --- E. Optional: Random missing values (uncomment if you want realism) ---
# for col in ['Sleep_Duration', 'Stress_Level']:
#     missing_idx = df.sample(frac=0.02).index
#     df.loc[missing_idx, col] = np.nan


# --- 8. Order columns ---
column_order = [
    'Age', 'Stress_Level', 'Gender', 'Menstruation', 'Sleep_Duration',
    'Pain_Severity', 'Light_Sensitivity', 'Noise_Sensitivity',
    'No_of_Meals', 'Water_Intake', 'Weather_Sunny', 'Weather_Cloudy',
    'Weather_Rainy', 'Weather_Snowy', 'Migraine_Score'
]
df = df[column_order]

# --- 9. Save ---
df.to_csv(FILE_NAME, index=False)
print(f"✅ Success! The dataset '{FILE_NAME}' with {N_ROWS} rows has been generated.")
print("\n--- Sample Data ---")
print(df.head(10))
```
